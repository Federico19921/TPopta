```python
import pandas as pd
```

```python
archivo = 'C:/Users/Federico/Desktop/TP Opta/data/raw/ACLED Data_2026-10-08_event_date_from_2021-10-01_event_date_to_2022-12-30 (1).csv'
```

```python
df = pd.read_csv(archivo)
```

```python
datos_chubut = df[df['admin1'] == 'Chubut']
print(datos_chubut)
```

```python
datos_chubut = datos_chubut[datos_chubut["event_date"] >= "2021-11-03"]
datos_chubut = datos_chubut[datos_chubut["event_date"] <= "2022-01-06"]
```

```python
print(datos_chubut)
```

```python
datos_chubut.to_excel("eventos_chubut_acled.xlsx", index=False)

```

```python
import pandas as pd

print("=== RESUMEN GENERAL ===")
print(f"Total eventos: {len(datos_chubut)}")
print(f"Columnas disponibles: {list(datos_chubut.columns)}\n")

print("=== POR TIPO DE EVENTO ===")
print(datos_chubut["event_type"].value_counts())
print()

print("=== POR SUBTIPO DE EVENTO ===")
print(datos_chubut["sub_event_type"].value_counts())
print()

print("=== REPRESIÓN ===")
represion = datos_chubut[
    datos_chubut["sub_event_type"].isin([
        "Protest with intervention",
        "Excessive force against protesters"
    ])
]
print(f"Eventos con represión (Protest with intervention + Excessive force): {len(represion)}")
print(f"Porcentaje del total: {100*len(represion)/len(datos_chubut):.1f}%\n")

print("=== DESGLOSE POR SUBTIPO ===")
print(represion["sub_event_type"].value_counts())
print()

print("=== REPRESIÓN POR PROVINCIA ===")
print(represion.groupby("admin1").size().sort_values(ascending=False))
print()

print("=== REPRESIÓN POR MES ===")
represion["fecha"] = pd.to_datetime(represion["event_date"])
represion["mes"] = represion["fecha"].dt.to_period("M")
print(represion.groupby(["admin1", "mes"]).size())
print()

print("=== ACTORES INVOLUCRADOS EN REPRESIÓN ===")
# Split por ; para ver actores individuales
actores = (
    represion["assoc_actor_1"]
    .dropna()
    .str.split(";")
    .explode()
    .str.strip()
)
print(actores.value_counts().head(20))
print()

print("=== MUESTRA DE EVENTOS CON REPRESIÓN ===")
print(represion[[
    "event_id_cnty", "event_date", "admin1", "location",
    "actor1", "assoc_actor_1", "notes"
]].head(10).to_string())
```
